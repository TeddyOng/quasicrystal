```python
import numpy as np
import matplotlib.pyplot as plt

# Define the potential function
def potential(x, y, V0, V1):
    l = 1
    k = 2 * np.pi / l
    
    V = V0 * (np.sin(k * x)**2 + np.sin(k * y)**2)
    V += V1 * (np.sin(k * (x + y) / np.sqrt(2))**2 + np.sin(k * (x - y) / np.sqrt(2))**2)
    
    return V

# Create a grid of x and y values
x = np.linspace(-2, 2, 500)
y = np.linspace(-2, 2, 500)
X, Y = np.meshgrid(x, y)

# Compute the potential over the grid
V0_value = 1.0  # Adjust as needed
V1_value = 1.0  # Adjust as needed
Z = potential(X, Y, V0_value, V1_value)

# Plotting the potential using a contour plot
plt.figure(figsize=(8, 6))
contour = plt.contourf(X, Y, Z, levels=5, cmap='viridis')
plt.contourf(X, Y, Z, levels=5, cmap='viridis', alpha=0.75)
plt.colorbar(contour, label='Potential V(x, y)')
plt.title('Potential Field')
plt.xlabel('x')
plt.ylabel('y')
plt.show()

```
