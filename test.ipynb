```python
import scipy.integrate
import numpy as np

def f(x, gamma):
    # print(f"gamma {gamma}")
    kappa = 1
    omega = 1
    return np.exp(- kappa * np.abs(x) - 1j * omega * x) * np.exp(- 1j * gamma * x)
```

```python
def IFF(f):
    return np.array([scipy.integrate.quad(f,-np.inf,np.inf,args=i, complex_func=True) for i in range(5000)], dtype=complex)
```

```python
IFF(f)
```
