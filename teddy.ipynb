```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def potential(V0: float, 
              V1: float, 
              x: float,
              y: float) -> float:
    """
    # Quasicrsytal potential 
    Adding a square grid with potential V0 
    and a diagonal grid with potential V1

    ## Inputs
    - V0: magnitude of potential for square grid
    - V1: magnitude of potential for diagonal grid
    - x: x-coordinate
    - y: y-coordinate

    ## Output
    - V(x,y) 
    """
    l = 1
    k = 2 * np.pi / l

    V = V0 * (np.sin(k*x)**2
            + np.sin(k*y)**2)
    
    V += V1 * (np.sin(k*(x+y)/np.sqrt(2))**2
             + np.sin(k*(x-y)/np.sqrt(2))**2)

    return V

# Create a grid of (x, y) values
x = np.linspace(-3, 3, 1000)
y = np.linspace(-3, 3, 1000)
X, Y = np.meshgrid(x, y)
Z = potential(1, 1, X, Y)


# Create the plot
plt.figure(figsize=(6,5))
contour = plt.contour(X, Y, Z, levels=5, cmap='viridis')  # Contour lines
plt.contourf(X, Y, Z, levels=5, cmap='viridis', alpha=0.75)  # Filled contours
plt.colorbar(label='V(x, y)')  # Add a color scale
plt.xlabel('x')
plt.ylabel('y')
plt.title('Color Plot of V(x,y)')
plt.show()
```
