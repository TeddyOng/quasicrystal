```python
import numpy as np
import matplotlib.pyplot as plt

def potential(V0, V1, x, y):
    """
    # Quasicrystal potential
    Adding a square grid with potential V0 and a diagonal grid with potential V1

    ## Inputs
    - V0: magnitude of potential for square grid
    - V1: magnitude of potential for diagonal grid
    - x: x-coordinate
    - y: y-coordinate

    ## Output
    - V(x,y)
    """
    
    l = 1
    k = 2 * np.pi / l

    V = V0 * (np.sin(k*x)**2 + np.sin(k*y)**2)
    V += V1 * (np.sin(k*(x+y)/np.sqrt(2))**2 + np.sin(k*(x-y)/np.sqrt(2))**2)

    return V


x = np.linspace(-3, 3, 1000)
y = np.linspace(-3, 3, 1000)
X, Y = np.meshgrid(x, y)
Z = potential(1, 1, X, Y)


plt.figure(figsize=(6,5))
plt.pcolormesh(X, Y, Z, shading='auto', cmap='viridis')
plt.colorbar(label='V(x, y)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Color Plot of V(x,y)')
plt.show()
```
