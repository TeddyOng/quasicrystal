```python
import numpy as np 
import matplotlib.pyplot as plt

def potential(x, y, V0, V1):
    l = 1
    k = 2 * np.pi /1

    V = V0 * (np.sin(k * x)**2 + np.sin(k * y)**2)
    V += V1 * (np.sin(k * (x + y) / np.sqrt(2))**2 + np.sin(k * (x - y) / np.sqrt(2))**2)

    return V

x = np.linspace(0, 2, 200)
y = np.linspace(0, 2, 200)
X, Y = np.meshgrid(x, y)

V0_value = 1.0
V1_value = 0.5
Z = potential(X, Y, V0_value, V1_value)

# Step 4: Plot the contour plot
plt.figure(figsize=(6, 6))
cp = plt.contour(X, Y, Z, 20)  # 20 contour levels
plt.colorbar(cp)  # Add a colorbar to show the potential values
plt.title("Contour plot of the potential")
plt.xlabel("x")
plt.ylabel("y")
plt.show()

```
